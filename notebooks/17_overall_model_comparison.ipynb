# Overall Comparison of Machine Learning Models

## Introduction

Previous experiments investigated the influence of different data quality factors on machine learning performance using the Titanic dataset.

Three machine learning algorithms were evaluated:

- Logistic Regression
- Decision Tree
- Random Forest

The experiments investigated three aspects of data quality:

- Missing Data
- Label Noise
- Feature Selection

The objective of this notebook is to compare the performance of all three machine learning models across these experiments and identify which algorithm is the most robust under different data quality conditions.

## Analysis of Results

The overall comparison demonstrates that the three machine learning models respond differently to changes in data quality.

For the Missing Data experiment, Random Forest achieved the highest Accuracy on the complete dataset, while Decision Tree showed the greatest robustness when moderate amounts of missing data were introduced. Logistic Regression experienced the largest decrease in performance as the proportion of missing values increased.

In the Label Noise experiment, all three models showed a substantial decline in performance as incorrect target labels were introduced. Logistic Regression and Random Forest experienced the largest performance degradation, while Decision Tree retained comparatively higher Accuracy and F1-score at higher noise levels.

For the Feature Selection experiment, Logistic Regression remained relatively stable across all feature configurations, indicating that additional features had only a limited influence on its performance. In contrast, both Decision Tree and Random Forest benefited considerably from richer feature sets, with Random Forest achieving the highest overall Accuracy (0.821) and F1-score (0.778).

Overall, the experiments indicate that different machine learning algorithms have different levels of robustness depending on the type of data quality issue. Tree-based algorithms generally performed better than Logistic Regression when additional informative features were available, while Decision Tree demonstrated greater robustness to label noise.

## Conclusion

This study compared the performance of Logistic Regression, Decision Tree, and Random Forest under three common data quality challenges: missing data, label noise, and feature selection.

The experiments demonstrated that no single algorithm performs best under every condition. Random Forest achieved the highest overall performance when high-quality data and informative features were available, while Decision Tree showed greater robustness when the dataset quality deteriorated.

Logistic Regression produced stable results during the feature selection experiment but was more sensitive to missing data and label noise than the tree-based algorithms.

These findings confirm that data quality has a significant influence on machine learning performance and that the choice of algorithm should consider the characteristics and quality of the available dataset. Proper preprocessing, accurate labels, and informative feature selection remain essential for building reliable predictive models.
